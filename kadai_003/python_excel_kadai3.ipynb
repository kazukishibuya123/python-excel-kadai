```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.styles import PatternFill

# =========================
# Excelファイルの読み込み
# =========================
file1 = "2022_年間売上表.xlsx"
file2 = "2023_年間売上表.xlsx"

df_2022 = pd.read_excel(file1, sheet_name="Sheet1")
df_2023 = pd.read_excel(file2, sheet_name="Sheet1")

# =========================
# データの連結
# =========================
df_all = pd.concat([df_2022, df_2023], ignore_index=True)

# =========================
# 商品・売上年ごとに集計
# =========================
df_summary = (
    df_all.groupby(["商品", "売上年"], as_index=False)["金額（千円）"]
    .sum()
)

# =========================
# Excel出力
# =========================
output_file = "売上集計表.xlsx"

df_summary.to_excel(
    output_file,
    sheet_name="集計結果",
    index=False
)

# =========================
# ヘッダーの色設定
# =========================
wb = load_workbook(output_file)
ws = wb["集計結果"]

header_fill = PatternFill(
    fill_type="solid",
    start_color="F2F2F2",
    end_color="F2F2F2"
)

for cell in ws[1]:
    cell.fill = header_fill

# 保存
wb.save(output_file)

print("売上集計表.xlsx を作成しました。")
```
