```python
from openpyxl import Workbook
from datetime import datetime

# Excelブック作成
wb = Workbook()
ws = wb.active

# 現在日付
today = datetime.now()

# ファイル名用
file_date = today.strftime("%Y%m%d")

# 表示用
display_date = today.strftime("%Y/%m/%d")

# タイトル
ws["B2"] = "請求書"

# 会社情報
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678"
ws["B7"] = "担当者名:鈴木一郎 様"

# 請求書情報
ws["F4"] = "No."
ws["G4"] = "0001"

ws["F5"] = "日付"
ws["G5"] = display_date

# ヘッダー
ws["B10"] = "商品名"
ws["C10"] = "数量"
ws["D10"] = "単価"
ws["E10"] = "金額"

# 商品データ
ws["B11"] = "商品A"
ws["C11"] = 2
ws["D11"] = 10000
ws["E11"] = 20000

ws["B12"] = "商品B"
ws["C12"] = 1
ws["D12"] = 15000
ws["E12"] = 15000

# 合計計算
subtotal = 35000
tax = 3500
total = 38500

ws["B15"] = "小計"
ws["E15"] = subtotal

ws["B16"] = "消費税"
ws["E16"] = tax

ws["B17"] = "合計"
ws["E17"] = total

# 保存
file_name = f"請求書_{file_date}.xlsx"
wb.save(file_name)

print(f"{file_name} を作成しました")
```
